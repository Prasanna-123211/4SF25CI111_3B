## 1. DataFrame with 3 columns and 5 rows; add product column

````python
# Create a DataFrame with 3 columns and 5 rows
df1 = pd.DataFrame(np.random.randint(1, 11, size=(5, 3)),
                   columns=["A", "B", "C"])

# Multiply the first two columns and store the result in a new column
df1["A_times_B"] = df1["A"] * df1["B"]

print(df1)
````

## 2. Row-wise and column-wise sums

````python
df2 = pd.DataFrame(np.random.randint(1, 21, size=(4, 3)),
                   columns=["A", "B", "C"])

print("DataFrame:")
print(df2)

print("\\nRow-wise sum:")
print(df2.sum(axis=1))  # axis=1 means across columns

print("\\nColumn-wise sum:")
print(df2.sum(axis=0))  # axis=0 means down rows
````

## 3. Replace NaN values with their column means

````python
df3 = pd.DataFrame(np.random.randint(1, 21, size=(5, 3)).astype(float),
                   columns=["A", "B", "C"])

# Introduce missing values
df3.iloc[1, 0] = np.nan
df3.iloc[3, 1] = np.nan
df3.iloc[4, 2] = np.nan

print("Before filling NaN values:")
print(df3)

# Each NaN is replaced by the mean of its own column
df3 = df3.fillna(df3.mean())

print("\\nAfter filling NaN values:")
print(df3)
````

## 4. Drop rows that contain any NaN value

````python
df4 = pd.DataFrame(np.random.randint(1, 51, size=(6, 4)).astype(float),
                   columns=["A", "B", "C", "D"])

# Introduce missing values in a few rows
df4.iloc[1, 2] = np.nan
df4.iloc[4, 0] = np.nan

print("Before dropping rows:")
print(df4)

df4_clean = df4.dropna()

print("\\nAfter dropping rows containing any NaN:")
print(df4_clean)
````

## 5. Group by Category and calculate sum and mean

````python
df5 = pd.DataFrame({
    "Category": np.random.choice(["A", "B", "C"], size=12),
    "Value": np.random.randint(10, 101, size=12)
})

print("Original DataFrame:")
print(df5)

summary5 = df5.groupby("Category")["Value"].agg(["sum", "mean"])
print("\\nSum and mean for each category:")
print(summary5)
````

## 6. Group products by Category and calculate total Sales

````python
df6 = pd.DataFrame({
    "Product": np.random.choice(["Pen", "Book", "Bag", "Bottle"], size=10),
    "Category": np.random.choice(["Stationery", "Accessories", "Books"], size=10),
    "Sales": np.random.randint(100, 1001, size=10)
})

print("Original DataFrame:")
print(df6)

total_sales = df6.groupby("Category")["Sales"].sum()
print("\\nTotal sales for each category:")
print(total_sales)
````

## 7. Merge two DataFrames using a common column

````python
left = pd.DataFrame({
    "StudentID": [1, 2, 3, 4],
    "Name": ["Asha", "Ravi", "Meena", "John"]
})

right = pd.DataFrame({
    "StudentID": [1, 2, 3, 4],
    "Marks": [85, 90, 78, 92]
})

merged = pd.merge(left, right, on="StudentID")
print("First DataFrame:")
print(left)
print("\\nSecond DataFrame:")
print(right)
print("\\nMerged DataFrame:")
print(merged)
````

## 8. MultiIndex DataFrame: indexing and slicing

````python
index = pd.MultiIndex.from_tuples(
    [("North", "2024"), ("North", "2025"),
     ("South", "2024"), ("South", "2025"),
     ("West", "2024"), ("West", "2025")],
    names=["Region", "Year"]
)

df8 = pd.DataFrame(
    {"Sales": [120, 150, 100, 130, 90, 110],
     "Profit": [20, 30, 15, 25, 10, 18]},
    index=index
)

print("MultiIndex DataFrame:")
print(df8)

print("\\nSelect all rows for North:")
print(df8.loc["North"])

print("\\nSelect the row for South in 2025:")
print(df8.loc[("South", "2025")])

print("\\nSelect 2024 rows from all regions:")
print(df8.xs("2024", level="Year"))
````

## 9. MultiIndex with Category and SubCategory; sum values

````python
df9 = pd.DataFrame({
    "Category": np.random.choice(["Food", "Clothing", "Electronics"], size=12),
    "SubCategory": np.random.choice(["A", "B"], size=12),
    "Value": np.random.randint(1, 101, size=12)
})

# Set Category and SubCategory as a hierarchical index
multi9 = df9.set_index(["Category", "SubCategory"]).sort_index()

print("DataFrame with MultiIndex:")
print(multi9)

print("\\nSum for each Category and SubCategory:")
print(multi9.groupby(level=["Category", "SubCategory"])["Value"].sum())

print("\\nTotal sum for each Category:")
print(multi9.groupby(level="Category")["Value"].sum())
````

## 10. Apply a function that doubles every value

````python
df10 = pd.DataFrame(np.random.randint(1, 21, size=(5, 3)),
                    columns=["A", "B", "C"])

print("Original DataFrame:")
print(df10)

doubled = df10.apply(lambda column: column * 2)
print("\\nDataFrame after doubling every value:")
print(doubled)
````

## 11. Use a lambda function to add a row-wise sum column

````python
df11 = pd.DataFrame(np.random.randint(1, 21, size=(6, 3)),
                    columns=["A", "B", "C"])

# axis=1 applies the lambda function to each row
df11["Row_Sum"] = df11.apply(lambda row: row["A"] + row["B"] + row["C"], axis=1)

print(df11)
````

## 12. Convert all strings in a Series to uppercase

````python
text_series = pd.Series(["hello", "pandas", "python", "data science", "notebook"])

print("Original Series:")
print(text_series)

uppercase_series = text_series.str.upper()
print("\\nUppercase Series:")
print(uppercase_series)
````

## 13. Extract the first three characters of each string

````python
text_series2 = pd.Series(["elephant", "tiger", "pandas", "notebook", "python"])

print("Original Series:")
print(text_series2)

first_three = text_series2.str[:3]
print("\\nFirst three characters:")
print(first_three)
````
